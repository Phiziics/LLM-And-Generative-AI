# 11 FastAPI Backend

## Goal

This notebook turns the RiskRadar AI RAG pipeline into a local API.

So far, the project works inside notebooks.

Now we create reusable Python modules and a FastAPI backend.

The workflow is:

```text
notebook RAG functions
→ reusable src modules
→ FastAPI app
→ local API endpoints
→ test requests
```

The backend will support:

```text
health check
company list
evidence retrieval
grounded RAG answers
optional financial context
```
