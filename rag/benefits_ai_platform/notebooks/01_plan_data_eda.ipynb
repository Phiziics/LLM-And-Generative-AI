```python
# Import libraries

from pathlib import Path

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)
pd.set_option("display.max_colwidth", 120)
```

```python
# Define project paths

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

RAW_DIR = PROJECT_ROOT / "data" / "01_raw" / "cms_exchange_pufs"
PREPROCESSED_DIR = PROJECT_ROOT / "data" / "02_preprocessed"

PREPROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("RAW_DIR:", RAW_DIR)
print("PREPROCESSED_DIR:", PREPROCESSED_DIR)
```

```python
# Find all CSV files in the CMS Exchange PUF raw folder

csv_files = sorted(RAW_DIR.rglob("*.csv"))

print("CSV files found:", len(csv_files))

for i, file_path in enumerate(csv_files):
    print(i, file_path.relative_to(PROJECT_ROOT))
```

```python
# Auto-detect the main CMS PUF files

def find_file_by_keywords(files, keywords):
    matches = []

    for file_path in files:
        file_path_text = str(file_path).lower()

        if all(keyword.lower() in file_path_text for keyword in keywords):
            matches.append(file_path)

    return matches


plan_matches = find_file_by_keywords(csv_files, ["plan"])
benefit_matches = find_file_by_keywords(csv_files, ["benefit"])
rate_matches = find_file_by_keywords(csv_files, ["rate"])

print("Plan matches:")
for i, file_path in enumerate(plan_matches):
    print(i, file_path.relative_to(PROJECT_ROOT))

print("\nBenefit matches:")
for i, file_path in enumerate(benefit_matches):
    print(i, file_path.relative_to(PROJECT_ROOT))

print("\nRate matches:")
for i, file_path in enumerate(rate_matches):
    print(i, file_path.relative_to(PROJECT_ROOT))
```

```python
# Select the main plan, benefits, and rate files

plan_file = plan_matches[0]
benefits_file = benefit_matches[0]
rate_file = rate_matches[0]

print("Plan file:", plan_file.relative_to(PROJECT_ROOT))
print("Benefits file:", benefits_file.relative_to(PROJECT_ROOT))
print("Rate file:", rate_file.relative_to(PROJECT_ROOT))
```

```python
# Load the Plan Attributes file
# This is the main file for plan-level metadata.

plan_df = pd.read_csv(plan_file, low_memory=False)

print("Plan data shape:", plan_df.shape)
display(plan_df.head())
```

```python
# Inspect plan columns, data types, and missing values

plan_columns = pd.DataFrame({
    "column": plan_df.columns,
    "dtype": [plan_df[col].dtype for col in plan_df.columns],
    "missing_count": [plan_df[col].isna().sum() for col in plan_df.columns],
    "missing_pct": [(plan_df[col].isna().mean() * 100).round(2) for col in plan_df.columns],
})

display(plan_columns.sort_values("missing_pct", ascending=False))
```

```python
# Search for important columns by keyword

def search_columns(df, keywords):
    matches = []

    for col in df.columns:
        col_lower = col.lower()

        if any(keyword.lower() in col_lower for keyword in keywords):
            matches.append(col)

    return matches


search_terms = {
    "plan_id": ["planid", "standardcomponentid"],
    "plan_name": ["planmarketingname", "marketing"],
    "issuer": ["issuer"],
    "state": ["state"],
    "metal": ["metal"],
    "plan_type": ["plantype", "type"],
    "deductible": ["deductible", "deduct"],
    "out_of_pocket": ["outofpocket", "out of pocket", "moop", "maximum"],
    "hsa": ["hsa"],
    "network": ["network"],
}

for label, keywords in search_terms.items():
    print(f"\n{label.upper()}")
    print(search_columns(plan_df, keywords))
```

```python
# Map CMS columns to clean project column names
# Note: CMS 2026 Plan Attributes uses IssuerMarketPlaceMarketingName instead of IssuerName.

def get_first_existing_column(df, candidates):
    for col in candidates:
        if col in df.columns:
            return col

    return None


column_map = {
    "plan_id": get_first_existing_column(
        plan_df,
        [
            "StandardComponentId",
            "PlanId",
        ],
    ),
    "issuer_id": get_first_existing_column(
        plan_df,
        [
            "IssuerId",
        ],
    ),
    "issuer_name": get_first_existing_column(
        plan_df,
        [
            "IssuerMarketPlaceMarketingName",
            "IssuerName",
        ],
    ),
    "state": get_first_existing_column(
        plan_df,
        [
            "StateCode",
            "State",
        ],
    ),
    "plan_name": get_first_existing_column(
        plan_df,
        [
            "PlanMarketingName",
            "PlanVariantMarketingName",
        ],
    ),
    "metal_level": get_first_existing_column(
        plan_df,
        [
            "MetalLevel",
        ],
    ),
    "plan_type": get_first_existing_column(
        plan_df,
        [
            "PlanType",
        ],
    ),
    "hsa_eligible": get_first_existing_column(
        plan_df,
        [
            "IsHSAEligible",
        ],
    ),
    "deductible_integrated": get_first_existing_column(
        plan_df,
        [
            "MedicalDrugDeductiblesIntegrated",
        ],
    ),
    "oop_integrated": get_first_existing_column(
        plan_df,
        [
            "MedicalDrugMaximumOutofPocketIntegrated",
        ],
    ),
    "individual_moop": get_first_existing_column(
        plan_df,
        [
            "MEHBInnTier1IndividualMOOP",
            "TEHBInnTier1IndividualMOOP",
        ],
    ),
}

column_map
```

```python
# Build clean plan dataframe from selected columns

clean_plan_df = pd.DataFrame()

for clean_col, source_col in column_map.items():
    if source_col is None:
        clean_plan_df[clean_col] = np.nan
    else:
        clean_plan_df[clean_col] = plan_df[source_col]

print("Clean plan shape:", clean_plan_df.shape)
display(clean_plan_df.head())
```

```python
# Clean text columns

text_cols = [
    "plan_id",
    "issuer_id",
    "issuer_name",
    "state",
    "plan_name",
    "metal_level",
    "plan_type",
    "hsa_eligible",
    "deductible_integrated",
    "oop_integrated",
]

for col in text_cols:
    clean_plan_df[col] = (
        clean_plan_df[col]
        .astype("string")
        .str.strip()
        .replace({
            "": pd.NA,
            "nan": pd.NA,
            "None": pd.NA,
            "Not Applicable": pd.NA,
        })
    )

display(clean_plan_df.head())
```

```python
# Clean numeric out-of-pocket field
# The deductible_integrated and oop_integrated fields are design indicators, not dollar amounts.

def clean_money(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value.lower() in ["", "nan", "none", "not applicable", "not available"]:
        return np.nan

    if "no charge" in value.lower():
        return 0.0

    value = (
        value
        .replace("$", "")
        .replace(",", "")
        .replace("%", "")
        .strip()
    )

    try:
        return float(value)
    except ValueError:
        return np.nan


clean_plan_df["out_of_pocket_individual"] = clean_plan_df["individual_moop"].apply(clean_money)

display(
    clean_plan_df[
        [
            "deductible_integrated",
            "oop_integrated",
            "individual_moop",
            "out_of_pocket_individual",
        ]
    ].head()
)

display(clean_plan_df[["out_of_pocket_individual"]].describe())
```

```python
# Check missing values after cleaning

missing_summary = clean_plan_df.isna().sum().reset_index()
missing_summary.columns = ["column", "missing_count"]
missing_summary["missing_pct"] = (missing_summary["missing_count"] / len(clean_plan_df) * 100).round(2)

display(missing_summary.sort_values("missing_pct", ascending=False))
```

```python
# Remove rows missing core plan identity fields only
# Do not remove rows because benefit design fields are missing.

before_rows = len(clean_plan_df)

required_cols = [
    "plan_id",
    "plan_name",
    "issuer_name",
    "state",
]

clean_plan_df = clean_plan_df.dropna(subset=required_cols).copy()

after_rows = len(clean_plan_df)

print("Rows before:", before_rows)
print("Rows after:", after_rows)
print("Rows removed:", before_rows - after_rows)

display(clean_plan_df.head())
```

```python
# Remove duplicate plan records

duplicate_count = clean_plan_df.duplicated(subset=["plan_id"]).sum()

print("Duplicate plan_id count:", duplicate_count)

clean_plan_df = clean_plan_df.drop_duplicates(subset=["plan_id"]).reset_index(drop=True)

print("Final shape:", clean_plan_df.shape)
```

```python
# Basic plan data summary

print("Number of plans:", clean_plan_df["plan_id"].nunique())
print("Number of issuers:", clean_plan_df["issuer_name"].nunique())
print("Number of states:", clean_plan_df["state"].nunique())

display(clean_plan_df["metal_level"].value_counts(dropna=False).to_frame("count"))
display(clean_plan_df["plan_type"].value_counts(dropna=False).to_frame("count"))
display(clean_plan_df["state"].value_counts(dropna=False).to_frame("count").head(20))
```

```python
# Numeric summary for out-of-pocket values

display(
    clean_plan_df[
        [
            "out_of_pocket_individual",
        ]
    ].describe()
)
```

```python
# Preview plans with available out-of-pocket data

plans_with_oop = (
    clean_plan_df
    .dropna(subset=["out_of_pocket_individual"])
    .sort_values("out_of_pocket_individual")
    [
        [
            "plan_name",
            "issuer_name",
            "state",
            "metal_level",
            "plan_type",
            "deductible_integrated",
            "oop_integrated",
            "out_of_pocket_individual",
        ]
    ]
    .head(20)
)

display(plans_with_oop)
```

```python
# Create readable values for RAG text

rag_df = clean_plan_df.copy()

rag_df["hsa_eligible_text"] = rag_df["hsa_eligible"].fillna("Not available")
rag_df["deductible_integrated_text"] = rag_df["deductible_integrated"].fillna("Not available")
rag_df["oop_integrated_text"] = rag_df["oop_integrated"].fillna("Not available")

rag_df["out_of_pocket_individual_text"] = rag_df["out_of_pocket_individual"].apply(
    lambda x: "Not available" if pd.isna(x) else f"${x:,.0f}"
)

display(
    rag_df[
        [
            "plan_name",
            "hsa_eligible_text",
            "deductible_integrated_text",
            "oop_integrated_text",
            "out_of_pocket_individual_text",
        ]
    ].head()
)
```

```python
# Build RAG text for each plan

def build_plan_summary(row):
    return f"""
Plan Name: {row["plan_name"]}
Plan ID: {row["plan_id"]}
Issuer: {row["issuer_name"]}
State: {row["state"]}
Metal Level: {row["metal_level"]}
Plan Type: {row["plan_type"]}
HSA Eligible: {row["hsa_eligible_text"]}

Deductible Design:
Medical and drug deductibles integrated: {row["deductible_integrated_text"]}.

Out-of-Pocket Design:
Medical and drug maximum out-of-pocket integrated: {row["oop_integrated_text"]}.
The individual in-network tier 1 maximum out-of-pocket amount is {row["out_of_pocket_individual_text"]}.

Benefits Context:
This plan record comes from public CMS Exchange Public Use File data.
It can be used for benefits search, plan comparison, document retrieval, and RAG question answering.
""".strip()


clean_plan_df["rag_text"] = rag_df.apply(build_plan_summary, axis=1)

display(clean_plan_df[["plan_id", "plan_name", "rag_text"]].head())
```

```python
# Save clean plan dataset

clean_output_path = PREPROCESSED_DIR / "clean_plan_data.csv"

clean_plan_df.to_csv(clean_output_path, index=False)

print("Saved:", clean_output_path)
print("Shape:", clean_plan_df.shape)
```

```python
# Create RAG-ready full and sample datasets

rag_ready_df = clean_plan_df[
    [
        "plan_id",
        "issuer_id",
        "issuer_name",
        "state",
        "plan_name",
        "metal_level",
        "plan_type",
        "hsa_eligible",
        "deductible_integrated",
        "oop_integrated",
        "out_of_pocket_individual",
        "rag_text",
    ]
].copy()

rag_ready_sample_df = rag_ready_df.head(1000).copy()

print("RAG-ready full shape:", rag_ready_df.shape)
print("RAG-ready sample shape:", rag_ready_sample_df.shape)

display(rag_ready_sample_df.head())
```

```python
# Save RAG-ready datasets

rag_full_output_path = PREPROCESSED_DIR / "rag_ready_plan_data_full.csv"
rag_sample_output_path = PREPROCESSED_DIR / "rag_ready_plan_data_sample.csv"

rag_ready_df.to_csv(rag_full_output_path, index=False)
rag_ready_sample_df.to_csv(rag_sample_output_path, index=False)

print("Saved:", rag_full_output_path)
print("Saved:", rag_sample_output_path)
```

```python
# Reload saved files to confirm they work

check_clean_df = pd.read_csv(clean_output_path)
check_rag_sample_df = pd.read_csv(rag_sample_output_path)

print("Clean data shape:", check_clean_df.shape)
print("RAG sample shape:", check_rag_sample_df.shape)

display(check_rag_sample_df.head())
```
